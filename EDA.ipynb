```python
# Import libraries for data visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Distribution of traffic volume
plt.figure(figsize=(8,5))
sns.histplot(df['traffic_volume'], bins=20, kde=True)
plt.title('Distribution of Traffic Volume')
plt.xlabel('Traffic Volume')
plt.ylabel('Frequency')
plt.show()
```

```python
# Distribution of average vehicle speed
plt.figure(figsize=(8,5))
sns.histplot(df['avg_vehicle_speed'], bins=20, kde=True)
plt.title('Distribution of Average Vehicle Speed')
plt.xlabel('Average Vehicle Speed (km/h)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Relationship between traffic volume and average vehicle speed
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='traffic_volume', y='avg_vehicle_speed')
plt.title('Traffic Volume vs Average Vehicle Speed')
plt.xlabel('Traffic Volume')
plt.ylabel('Average Vehicle Speed (km/h)')
plt.show()
```

```python
# Relationship between weather condition and traffic volume
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='weather_condition', y='traffic_volume')
plt.title('Weather Condition vs Traffic Volume')
plt.xlabel('Weather Condition')
plt.ylabel('Traffic Volume')
plt.show()
```

```python
# Relationship between temperature and traffic volume
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='temperature', y='traffic_volume')
plt.title('Temperature vs Traffic Volume')
plt.xlabel('Temperature')
plt.ylabel('Traffic Volume')
plt.show()
```

```python
# Relationship between humidity and traffic volume
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='humidity', y='traffic_volume')
plt.title('Humidity vs Traffic Volume')
plt.xlabel('Humidity')
plt.ylabel('Traffic Volume')
plt.show()
```

```python
# Traffic volume based on accident reports
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='accident_reported', y='traffic_volume')
plt.title('Accident Reported vs Traffic Volume')
plt.xlabel('Accident Reported (0 = No, 1 = Yes)')
plt.ylabel('Traffic Volume')
plt.show()
```

```python
# Correlation between numerical features
plt.figure(figsize=(10,7))
numeric_df = df.select_dtypes(include='number')
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```
